# LIDO-Auswertungen

## Zahl der LIDO-Records

> Schreibe ein Python-Snippet, das die LIDO-XML-Datei(en) im Verzeichnis `lidos/` ausliest und die Zahl der <lido:lido>-Elemente ausgibt. Verwende lxml.etree.

## Zahl der Events je Record

> Schreibe ein Python-Snippet, das die Zahl der <lido:event>-Elemente je <lido:lido>-Element ausgibt. Die LIDO-XML-Datei(en) liegen im Verzeichnis `lidos/`. Verwende lxml.etree.

## ActorIDs

> Schreibe ein Python-Snippet, das die Werte aller <lido:actorID>-Elemente in eine Liste schreibt. Die LIDO-XML-Datei(en) liegen im Verzeichnis lidos/. Verwende lxml.etree.
